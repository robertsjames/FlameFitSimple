```python
import matplotlib.pyplot as plt
```

```python
exp_LNGS = [162.11, 200, 500.52]
exp_SURF = [189.95, 665.90]

sens_LNGS = [2.21e-49, 1.99e-49, 1.05e-49]
sens_SURF = [2.17e-49, 9.61e-50]


plt.figure(figsize=(7,5))

plt.scatter(exp_LNGS, sens_LNGS, color='r', label="LNGS")

plt.scatter(exp_SURF, sens_SURF, color='b', label="SURF")

plt.ylabel("Cross section [cm^2]", fontsize=12)
plt.xlabel("Exposure [ty]", fontsize=12)

plt.legend(fontsize=12, loc='upper right')
plt.show()
```
